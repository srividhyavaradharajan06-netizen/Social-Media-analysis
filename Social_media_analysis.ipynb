```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("data.csv")
df.head(5)
```

```python
df.shape
```

```python
df.info()
```

```python
df['posted_at']=pd.to_datetime(df['posted_at'])
df.info()
```

```python
categorical=df.select_dtypes('O').columns
for i in categorical:
  print(f"{df[i].isnull().value_counts()}\n")
```

```python
numerical=df.select_dtypes('number').columns
for i in numerical:
  print(f"{df[i].isnull().value_counts()}\n")
```

```python
numerical=df.select_dtypes('number').columns
for i in numerical:
  df[i].fillna(df[i].median(),inplace=True)
```

```python
for i in numerical:
  print(f"{df[i].isnull().value_counts()}\n")
```

```python
for i in categorical:
  df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
for i in categorical:
  print(f"{df[i].isnull().value_counts()}\n")
```

```python
df['engagement_score'] = df['likes'] + df['comments'] + df['shares']
df['engagement_score'].head(10)
# This creates a new column 'engagement_score' by combining likes, comments and shares.
# It represents the overall interaction a post receives as a single metric.
# Higher engagement score means the post generated more total user interaction.
```

```python
df['hashtag_count'] = df['hashtags'].apply(lambda x: len(str(x).split('#')) - 1)
df['hashtag_count'].head(7)
# This extracts the number of hashtags used in each post by counting the '#' symbol.
# str(x) converts any non-string values to string to avoid errors.
# Subtracting 1 removes the extra count caused by splitting before the first '#'.
```

```python
df.groupby('post_type')['engagement_score'].mean()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['likes'], df['impression_count'],alpha=0.2, s=8, color='steelblue')
plt.xlabel('Likes')
plt.ylabel('Impression Count')
plt.title('Likes vs Impression Count')
plt.show()
# This scatter plot examines the relationship between likes and impression count.
# It helps identify whether posts with more likes also tend to have higher impressions.
# Each dot represents a single post, revealing patterns or correlations between visibility and engagement.
```

```python
#line plot
df['posted_at']=pd.to_datetime(df['posted_at'])
df['month'] = df['posted_at'].dt.month
e1=df.groupby('month')['engagement_rate'].mean()
plt.plot(e1,color='blue',marker='>')
plt.xlabel('Month')
plt.ylabel('Avg Engagement Rate')
plt.title('Engagement Rate Over Time')
plt.show()
```

```python
uc = df.groupby('post_category')['user_id'].count()
plt.figure(figsize=(12, 5))
bars = plt.bar(uc.index, uc.values, color='black')
plt.bar_label(bars)
plt.xlabel('Post Category')
plt.ylabel('Number of Posts')
plt.title('Posts by Category')
plt.show()
# This bar chart shows the total number of posts in each content category.
# It helps identify which category has the most and least content on the platform.
# Useful for understanding what type of content users post the most.
```

```python
plt.pie(g.values,labels=g.index,autopct='%1.1f%%',colors=['red','blue','magenta'],explode=[0.1,0.1,0.1])
plt.title('Gender Distribution')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df['age'],bins=20,color='yellow',edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Age Distribution')
plt.show()
# This histogram shows the distribution of user ages across the dataset.
# It helps identify which age group has the highest number of users on the platform.
# The shape of the distribution reveals whether the platform is more popular among younger or older users.
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(x=df['post_type'],data=df)
plt.grid()
plt.xlabel('Post Type')
plt.ylabel('Count')
plt.title('Posts by Type')
plt.show()
```

```python
avg_likes = df.groupby('post_category')['likes'].mean().reset_index()
plt.figure(figsize=(12, 5))
ax = sns.barplot(data=avg_likes,x='post_category',y='likes',hue='post_category',legend=False,palette='Set2')
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f', fontsize=9, padding=3)

plt.xlabel('Post Category')
plt.ylabel('Average Likes')
plt.title('Average Likes by Category')
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.violinplot(x=df['sentiment'],y=df['follower_count'],hue='sentiment',legend=False,data=df,palette='Set2')
plt.xlabel('Sentiment')
plt.ylabel('Followers')
plt.title('Followers Distribution by Sentiment')
plt.show()
# This chart shows the distribution of follower counts across different sentiment categories.
# The width of each violin represents how many users have that follower count at each sentiment.
# It helps identify whether posts with positive, negative or neutral sentiment attract more followers.
```

```python
plt.figure(figsize=(12, 5))
sns.heatmap(correlation_matrix ,annot=True)
plt.title("Correlation Heatmap")
plt.show()
# This heatmap displays the correlation between all numerical variables in the dataset.
# Values close to 1 indicate strong positive correlation, -1 indicates strong negative correlation
# and values close to 0 indicate no relationship between the variables.
```

```python
country_eng = df.groupby('country')['engagement_rate'].mean().reset_index()
plt.figure(figsize=(14, 6))
ax = sns.barplot(data=country_eng, x='country', y='engagement_rate',hue='country',legend=False,palette='Set2')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', fontsize=8, padding=3)

plt.xlabel('Country')
plt.ylabel('Average Engagement Rate')
plt.title('Average Engagement Rate by Country')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
ax = sns.scatterplot(data=df, x='age', y='engagement_rate',hue='post_type',alpha=0.5,palette='dark')
plt.xlabel('Age')
plt.ylabel('Engagement Rate')
plt.title('Age vs Engagement Rate')
plt.show()
# This chart explores the relationship between user age and engagement rate across different post types.
# It helps identify which age groups are more engaged and whether post type influences engagement behavior.
# Each color represents a different post type, revealing which content resonates with which age group.
```

```python
ver_eng = df.groupby('is_verified')['engagement_rate'].mean().reset_index()

plt.figure(figsize=(8, 5))
ax = sns.barplot(data=ver_eng, x='is_verified', y='engagement_rate',hue='is_verified', legend=False,palette='Set2')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', fontsize=10, padding=3)

plt.xlabel('Is Verified')
plt.ylabel('Avg Engagement Rate')
plt.title('Verified vs Non-Verified Engagement Rate')
plt.show()
# This chart compares the average engagement rate between verified and non-verified users.
# It helps identify whether account verification has any impact on user engagement.
# Higher bar indicates that group receives more likes, comments and shares on average.
```
